## Q1: One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder

words = [["cat"], ["dog"], ["bird"], ["cat"]]

encoder = OneHotEncoder()

encoded = encoder.fit_transform(words)

print(encoded.toarray())
```

## Q2: TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love NLP",
    "NLP is amazing",
    "Machine learning is fun"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.toarray())
```

## Q3: CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["apple orange apple", "orange banana"]

vectorizer = CountVectorizer()

matrix = vectorizer.fit_transform(docs)

print(matrix.toarray())
```

```python
from gensim.models import Word2Vec

sentences = [
    ["I", "love", "NLP"],
    ["NLP", "is", "awesome"],
    ["Machine", "learning", "is", "fun"]
]

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1
)

print(model.wv["NLP"])
```

## Q7: Remove Numbers

## Q8: Extract Email

```python
import re

text = """
Contact:
student@gmail.com
teacher@yahoo.com
"""

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

print(emails)
```

## Q9: Remove Extra Spaces

## Q10: Tokenization

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

text = "Natural Language Processing is amazing."

tokens = word_tokenize(text)

print(tokens)
```

## Q11: Stopword Removal

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

words = ["I", "love", "to", "learn", "NLP"]

stop_words = set(stopwords.words("english"))

filtered = [word for word in words if word.lower() not in stop_words]

print(filtered)
```

## Q12: Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays"]

stems = [stemmer.stem(word) for word in words]

print(stems)
```

## Q13: Lemmatization

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]

lemmas = [lemmatizer.lemmatize(word) for word in words]

print(lemmas)
```

## Q14: Remove URLs

## Q16: Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love NLP",
    "NLP is fun"
]

#apply bag of words
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

## Q17: Access Word2Vec Vector

## Q18: Regex Phone

```python
import re

text = "Call me at +8801712345678"

phone = re.findall(
    r"\+\d{13}",
    text
)

print(phone)
```

## Q19: Remove Special Characters

```python
import re

text = "Hello@#$%^World!"

clean = re.sub(
    r"[^A-Za-z0-9\s]",
    "",
    text
)

print(clean)
```

## Q20: Complete NLP Preprocessing

```python
import re

text = "Hello!!! Visit https://abc.com Email: student@gmail.com 123"

text = text.lower()

text = re.sub(r"https?://\S+", "", text)
text = re.sub(r"\S+@\S+", "", text)
text = re.sub(r"\d+", "", text)
text = re.sub(r"[^\w\s]", "", text)
text = re.sub(r"\s+", " ", text).strip()

print(text)
```
